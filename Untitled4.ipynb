```python
import random
# 1st experiment
tosses = [random.choice([0,1]) for _ in range(100)]
num_heads = tosses.count(1)
num_tails = tosses.count(0)
print ("number of heads",num_heads)
print ("number of tails",num_tails)
# probablity = favourable outcomes/ total number of outcomes
print ("probablity of heads ",num_heads/100)
print ("probablit of tails",(num_tails)/100)
```

```python
#experiment 2
tosses = [random.choice([0,1]) for _ in range(1000)]
num_heads = tosses.count(1)
num_tails = tosses.count(0)
print ("number of heads",num_heads)
print ("number of tails",num_tails)
# probablity = favourable outcomes/ total number of outcomes
print ("probablity of heads ",num_heads/1000)
print ("probablit of tails",(num_tails)/1000)
# when we increase number of times coin tossed , we notice that difference between probablity of heads and tails starts getting reduced
```

```python
#experiment 4
rolls = [random.randint(1,6) for _ in range(10000)]
# radnom.randint select one number randomly between 1 and 6
num_1=rolls.count(1)
num_6=rolls.count(6)
prob_1 = (num_1/10000)
prob_6 = (num_6/10000)
prob_of_neither = (1-prob_1)*(1-prob_6)
# probablity formula to get probablity of neither
prob_of_both = prob_1*prob_6
if (prob_1 + prob_6==prob_of_both):
  print("P(AUB)==P(A)+P(B) is true" )

else:
  print("P(AUB)==P(A)+P(B) is false" )
```

```python
#experiment 5
rolls = [random.randint(1,6) ]
count_even = rolls.count(2)+rolls.count(4)+rolls.count(6)
count_odd = rolls.count(1)+rolls.count(3)+rolls.count(5)
prob_even = count_even
prob_odd = 1-prob_even
prob_neither = (1-prob_even)*(1-prob_odd)
prob_both = 1-prob_neither
if (prob_both==prob_even + prob_odd - prob_neither):
   print("P(AUB)==P(A)+P(B)-P(A intersection B ) is true")
else:
    print("P(AUB)==P(A)+P(B)-P(A intersection B ) is false")
```
